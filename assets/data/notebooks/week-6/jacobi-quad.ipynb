```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import scipy.special as sp
```

```python
f = lambda x: 100 * (np.sin(2 * np.pi * x) * (1/((x-3)**2 + 0.1)))

n, N = 15, 500
X = np.linspace(-1, 1, N)
```

Wählt $(\alpha, \beta)$ für:
- $(0, 0):\quad$ Gauss-Legendre
- $(1, 0):\quad$ Gauss-Radau (rechts)
- $(0, 1):\quad$ Gauss-Radau (links)
- $(0, 0):\quad$ Gauss-Lobatto (beide Randpunkte)

```python
approx = []

alpha, beta = 0, 0
x, w = sp.roots_jacobi(n, alpha, beta)

approx.append(np.sum(f(x)*w))

plt.plot(X, f(X), color="gray")
plt.bar(x, f(x), width=2*w)
print(f"Approximated integral: {np.sum(w * f(x)):.3f}")

# time.sleep(0.3)
plt.show()
```
