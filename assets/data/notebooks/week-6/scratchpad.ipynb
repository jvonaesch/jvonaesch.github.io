```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
f = np.sin
g = np.exp
h = lambda x: 100 * (np.sin(2 * np.pi * x) * (1/((x-3)**2 + 0.1)))

n = 10
```

```python
c, w = scipy.special.roots_jacobi(n, 1, 1)
X = np.linspace(-1, 1, 500)

plt.plot(X, h(X))
plt.bar(c, h(c), width=w, color="cyan")
```

```python
print(f"Integral: {scipy.integrate.quad(h, -1, 1)[0]}")
print(np.sum(w * h(c)))
```
