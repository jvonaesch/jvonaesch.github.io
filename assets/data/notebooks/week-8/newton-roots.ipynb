```python
import numpy as np
import matplotlib.pyplot as plt
```

## Fixpunktiteration
Die folgende Funktion nimmt einen Vektor von Werten $x$, über die die Wurzel berechnet werden soll, und berechnet iterativ die Wurzel aller $x$ mit dem Newton-Verfahren. Die rekursive Formel lautet:

$y_{i+1}=\frac{y_i+\frac{x}{y_i}}{2}$

Dabei gilt:     $\lim_{i\to\infty}(y_i)=\sqrt{x}$

```python
def approx_sqrt(x, d):
    y = [1]
    for i in range(1, d):
        y.append((y[-1] + x/y[-1]) / 2)
    return np.array(y)
```

```python
d = 30
x = 389056

# berechne alle Schritte
plt.figure(dpi=150)
plt.title("Wurzel mit Newton-Fixpunktiteration")
plt.xlabel(r"Iterationsschritt $k$")
plt.ylabel(r"Fehler von $x^{(k)}$")
y = approx_sqrt(x, d)
print(f"Square root of {x} approximately {y[-1]}")
print(f"\n\t- {y[-1]}^2 = {y[-1]**2}")
error = np.abs(y - np.sqrt(x))
plt.semilogy(np.arange(d), error)
```
