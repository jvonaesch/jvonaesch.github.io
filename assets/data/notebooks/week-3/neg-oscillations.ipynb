```python
import numpy as np
import matplotlib.pyplot as plt
```

Lege einige Parameter für die Messung fest: 

`n` -    Anzahl Messstellen <br>
`t` -    Messstellen (Müssen äquidistant sein) <br>
`N` -    Anzahl Auswertungsstellen (Darstellung) <br>
`T` -    Auswertungsstellen (Darstellung)

```python
n = 10
N = 500
t = np.linspace(0, 1, n)
T = np.linspace(0, 1, N)
```

`Osc(k)` entspricht $O_k$ mit $O_k(t)=e^{2\pi ikt}$

```python
def Osc(k):
    return (lambda t: np.real(np.exp(2j * np.pi * k * t)))
```

Definiere den positiven und entsprechenden negativen Oszillator, werte an allen $T$ aus und plotte zusammen mit den Messstellen $t_k$. 

Dies zeigt, dass $O_{-k}$ und $O_{N-1-k}$ ausgewertet an Stellen $t_k=\frac{k}{N}$ äquivalent sind. 

```python
for k in range(n):
    O = Osc(-k)
    P = Osc(n-k-1)

    plt.title(r"$n=10, k=%s, n-k-1=%s$" % (k, n-k-1))
    plt.scatter(t, O(t))
    plt.plot(T, O(T), label=r"$Real(e^{2\pi i(-%s)t})$" % (k))
    plt.plot(T, P(T), label=r"$Real(e^{2\pi i%st})$" % (n-k-1))
    plt.gca().set_yticklabels([])
    plt.legend()
    plt.show()
```
