## Setup

```python
import numpy as np
from matplotlib import pyplot as plt
```

Definiere die Funktion $f$ und deren Auswertungsstelle $x_0$ (`x`)

```python
f = np.exp
df = np.exp
x = 0
```

Parameter für die Auswertung und Darstellung:

```python
base = 10
n = 20
h = (1 / base) ** np.arange(n)
```

## Approximation
Approximiere die Ableitung mithilfe des imaginären Schritts und des Differenzenquotienten

```python
approx = (f(x + h) - f(x)) / h
i_approx = (f(x + 1j*h) - f(x)).imag / h

# rel. Fehler
actual = df(x)
rel_error = abs((approx - actual) / actual)
i_rel_error = abs((i_approx - actual) / actual)
```

## Plot
plotte den relativen Fehler:

```python
plt.title("Relativer Fehler der Differenzenquotienten")
plt.loglog(h, h/2, color="#dddddd")
plt.loglog(h, rel_error, color="maroon")
plt.loglog(h, (h**2)/4, color="#eeeeee")
plt.loglog(h, i_rel_error, color="darkgreen")
plt.legend([r"$\frac{h}{2}$", "erster Differenzenquotient", r"$\frac{h^2}{4}$", "komplexer Differenzenquotient"])
plt.xlabel(r"$h$")
plt.show()
```
