## Setup
Die folgenden Module werdet Ihr fast immer gebrauchen: 

```python
import numpy as np
import matplotlib.pyplot as plt
```

Die folgenden Methoden verwenden wir oft, um Vektoren und Matrizen zu generieren (j ist die imaginäre Konstante). `dtype=complex` lässt sich bei all diesen verwenden. 

```python
a = np.zeros((4, 6)) 
print(f"Alles nullen:\n{a}")

b = np.ones((6, 1), dtype=complex)
print(f"Alles einsen (komplex, vertikaler Vektor):\n{b}")

c = np.array([              
    [5 + 7j, 6, 7],
    [8, 9, 0]
])
print(f"2D-Array zum Erstellen einer Matrix:\n{c}")

d = np.arange(6)
print(f"Integers von 0 bis 5:\n{d}")

e = np.linspace(0, 1, 4)
print(f"200 äquidistante Punkte zwischen 0 und 1:\n{e}")
```

## Operationen
Um Elemente eines Vektors zu setzen, gibt es folgende Notationen:

```python
# Ein Element setzen (n argumente für n dimensionen)
a[3, 3] = 1 
a[:, 2] = 2        # ':'  setzt alle elemente in dieser dimension
a[1, :2] = 3       # ':x' setzt alle  Elemente bis zum x-ten
a[2:, 4:] = 4      # 'y:' setzt alle nach einem index (kombinierbar zu 'y:x')
a[0, :4] = d[:4]   # Wir können einen ganzen Vektor als Teil einer Matrix einfügen

# Ganze sub-matrizen können zugewiesen werden. 
a[2:4, 0:2] = np.ones((2, 2))

print(a)
```

Das Symbol für eine Matrix-Multiplikation ist `@`, und Transposition ist das attribut `T`:

```python
print(a @ b)
```

```python
print(c.T)
```

Numpy hat die meisten wichtigen mathematischen Funktionen. Diese lassen sich auf ganze Vektoren komponentenweise anwenden:

```python
f = np.exp
print(f(d))
```

Auch die arithmetischen Operationen aus python lassen sich komponentenweise anwenden:

```python
print(b+1)
print(5**d)
```

Wenn die Dimensionen von Vektoren übereinstimmen, passiert die Operation direkt komponentenweise.
Hier berechnen wir $x + \sqrt{x}$ für alle Komponenten:

```python
print(d + np.sqrt(d))
```

Wenn eine Dimension in einem Vektor $1$ ist, und im anderen $n$, dann wird die eindimensionale Komponente auf alle $n$ Zeilen/Spalten angewendet:

```python
print(d ** np.array([e]).T)
```

Hier ist der Exponent (rechts) senkrecht, und d waagrecht, deswegen zeigt die Matrix jede komponente in $d$ (Zeile) hoch jede Komponente in $e^T$ (Spalte):

$[{d_i}^{e_j}]_{i=0, j=0}^{n, m}$
