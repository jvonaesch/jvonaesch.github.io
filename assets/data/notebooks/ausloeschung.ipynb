## Setup

```python
import numpy as np
from matplotlib import pyplot as plt
```

Definiere Funktion und deren Auswertungsstelle und Ableitung für Fehlerberechnung:

```python
f = np.exp
df = np.exp
x = 0
```

Variablen für Plot und Auswertungsstellen:

```python
base = 10
n = 20
h = (1 / base) ** np.arange(n)
```

## Approximation

```python
approx = (f(x + h) - f(x)) / h

# Berechne rel. Fehler
actual = df(x)
rel_error = abs((approx - actual) / actual)
```

## Plot

```python
plt.title("Relativer Fehler der Differenzenquotienten")
plt.loglog(h, h/2, color="#dddddd")
plt.loglog(h, rel_error, color="maroon")
plt.legend([r"$\frac{h}{2}$", "relativer Fehler"])
plt.xlabel(r"$h$")
plt.show()
```
