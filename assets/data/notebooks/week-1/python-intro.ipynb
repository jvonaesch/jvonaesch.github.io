## Operatoren

| Java | Python | Bemerkung |
|---|---|---|
| `7 / 2.0` → `3.5` | `7 / 2` → `3.5` | `/` liefert in Python **immer** einen `float` |

### Schleifen

**Java**
```java
for (int i = 0; i < 5; i++) { ... }          
for (String name : names) { ... }            
while (bedingung) { ... }
do { ... } while (bedingung);
```

**Python**
```python
for i in range(5): ...           # 0, 1, 2, 3, 4
for name in names: ...           # jede for-Schleife ist eine for-each-Schleife
while bedingung: ...
# do-while gibt es nicht → while True: ... if ...: break
```

`range(start, stop, step)` erzeugt die Zahlen von `start` bis `stop`, wobei `stop` selbst nicht mehr dazugehört. `break` und `continue` funktionieren wie in Java.

````python
for i in range(1, 5):
    print(i)
print()
````

Schritt

````python
for i in range(2, 11, 3):        
    print(i, end=" ")
print()
````

````python
namen = ["Ada", "Linus", "Grace"]
for name in namen:
    print(f"Hallo {name}")

# Index UND Element: enumerate statt manueller Zähler
for i, name in enumerate(namen):
    print(i, name)

# Zwei Listen parallel durchlaufen: zip
alter = [36, 54, 85]
for name, a in zip(namen, alter):
    print(f"{name} ist {a} Jahre alt")

# while mit break – das Python-Muster für do-while
n = 0
while True:
    n += 1
    if n >= 3:
        break
print("n =", n)

# Besonderheit: else an einer Schleife läuft, wenn kein break ausgelöst wurde
for zahl in [2, 4, 6]:
    if zahl % 2 == 1:
        print("ungerade Zahl gefunden")
        break
else:
    print("nur gerade Zahlen")
````

## Listen
Python bringt die wichtigsten Sammlungen direkt in der Sprache mit – ohne Import, mit eigener Literal-Syntax.

| Java | Python | Literal |
|---|---|---|
| `ArrayList<T>` | `list` | `[1, 2, 3]` |
| `int[]` (Array) | `list` (oder NumPy) | `[1, 2, 3]` |
| – | `tuple` (unveränderliche Liste) | `(1, 2, 3)` |
| `HashMap<K, V>` | `dict` | `{"a": 1, "b": 2}` |
| `HashSet<T>` | `set` | `{1, 2, 3}` |

Python-Listen dürfen gemischte Typen enthalten (`[1, "zwei", 3.0]`) – in der Praxis ist das aber selten sinnvoll.

### Listen (≈ `ArrayList`)

| Java | Python |
|---|---|
| `list.add(x)` | `liste.append(x)` |
| `list.add(0, x)` | `liste.insert(0, x)` |
| `list.get(i)` | `liste[i]` |
| `list.set(i, x)` | `liste[i] = x` |
| `list.remove(Object)` | `liste.remove(x)` |
| `list.remove(int)` | `liste.pop(i)` oder `del liste[i]` |
| `list.size()` | `len(liste)` |
| `list.contains(x)` | `x in liste` |
| `Collections.sort(list)` | `liste.sort()` (in-place) oder `sorted(liste)` (neue Liste) |
| `list.subList(a, b)` | `liste[a:b]` |

````python
zahlen = [3, 1, 4, 1, 5]
zahlen.append(9)
zahlen.insert(0, 2)
print(zahlen, len(zahlen))

print(zahlen[0], zahlen[-1], zahlen[1:4])   # -1 = letztes Element, [1:4] = Slice mit Index 1 bis 3
print(4 in zahlen)

zahlen.sort()                          # sortiert die Liste selbst
print(zahlen)
print(sorted(zahlen, reverse=True))    # neue, absteigend sortierte Liste
````

- `def` statt Rückgabetyp und Modifikatoren; Typen sind nicht nötig (optional als Hinweise: `def f(x: float) -> float:`).
- Funktionen brauchen keine Klasse – sie stehen einfach auf oberster Ebene.
- Kein Überladen (mehrere Funktionen gleichen Namens).
- Eine Funktion ohne `return` liefert `None`.
- Mehrere Rückgabewerte: einfach ein Tupel zurückgeben und beim Aufruf entpacken.

````python
def returns_int() -> int:
    return 3.2

returns_int() 
# Lüge!
````

````python
# Standardwerte statt Überladen
def begruessung(name, gruss="Hallo", ausrufezeichen=True):
    text = f"{gruss} {name}" # f strings
    return text + "!" if ausrufezeichen else text

print(begruessung("Ada"))
print(begruessung("Ada", "Guten Tag"))
print(begruessung("Ada", ausrufezeichen=False))    # Keyword-Argument: Reihenfolge egal
````

````python
# Beliebig viele Argumente 
def summe(*werte):
    return sum(werte)

print(summe(1, 2, 3, 4))
````

### Indizieren, Slicing und Masken

Slicing funktioniert wie bei Listen. Zusätzlich kann man mit einem booleschen Array filtern – das ersetzt viele «`if` in einer Schleife»-Konstrukte.

````python
zahlen = np.array([1, 2, 3, 4, 5])

print(zahlen[0], zahlen[-1], zahlen[2:5])

maske = zahlen > 2          # boolesches Array
print(maske)
print(zahlen[maske])         # nur die Werte, bei denen die Maske True ist
print(zahlen[zahlen > 2].mean())


# Bedingte Zuweisung: alle Werte > 2 auf 0 setzen (Kopie, damit zahlen unverändert bleibt)
kopie = zahlen.copy()
kopie[kopie > 2] = 0
print(kopie)
````

Äusseres Produkt (Multiplikationstabelle) ohne meshgrid:

Meshgrid-Beispiel

````python
import numpy as np
import matplotlib.pyplot as plt


def huegel(x, y):
    """Ein Hügel: Höhe 1 im Ursprung, fällt nach aussen ab."""
    return np.exp(-(x**2 + y**2))


# 1) Achsen: 5 Punkte von -2 bis 2
x = np.linspace(-2, 2, 5) # [-2. -1.  0.  1.  2.]
y = np.linspace(-2, 2, 5)

# 2) Gitter: alle 5 x 5 = 25 Kombinationen von x und y
X, Y = np.meshgrid(x, y)

# 3) Funktion auf allen 25 Punkten gleichzeitig auswerten
Z = huegel(X, Y)

print("X =\n", X)
print("Y =\n", Y)
print("Z =\n", np.round(Z, 2))       # Zeile i, Spalte j: Höhe am Punkt (x[j], y[i])

plt.figure()
plt.contourf(X, Y, Z)
plt.colorbar()
plt.title("Hügel von oben")

plt.show()
````
