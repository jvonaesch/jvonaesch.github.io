## Setup

```python
import sympy

x, y = sympy.symbols(("x", "y"))
f_expr = sympy.exp(x) + y
```

## Auswertungen
Sympy expressions lassen sich mit expliziten Symbolzuweisungen auswerten: 

```python
result = f_expr.evalf(subs={
    x: 0,
    y: 1
})
print(f"f Direkt: \t{result}")
```

Oft ist es allerdings praktischer, eine expression zu einem lambda umzuwandeln

```python
f = sympy.lambdify((x, y), f_expr, "numpy")
print(f"f Lambdify: \t{f(0, 1)}")
```

Dies ist besonders hilfreich, um analytische Ableitungen und Integrale zu finden:

```python
# Ableitung
df_expr = sympy.diff(f_expr, x)
df = sympy.lambdify((x, y), df_expr, "numpy")
print(f"f Ableitung: \t\t{df(0, 1)}")
print(f"f Formel Ableitung: \t{df_expr}")

# Integration
g_expr = x
G_expr = sympy.integrate(x)
G = sympy.lambdify(x, G_expr, "numpy")
print(f"g Integral 0 bis 1: \t{G(1) - G(0)}")
```
