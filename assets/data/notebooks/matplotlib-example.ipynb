## Setup
Importiere matplotlib und numpy; diese Imports werdet ihr fast in jeder Aufgabe sehen. 

```python
import matplotlib.pyplot as plt
import numpy as np
```

Nun definieren wir einige Funktionen und einige Auswertungsstellen mithilfe von Numpy:

```python
f = np.exp
g = np.sin
h = np.log

x = np.linspace(0.1, 10, 500, endpoint=True) # 500 Stellen zwischen 0 und 3 inkl. Endpunkt
y = f(x)
```

## Plot
Einfache Plots könnt ihr in einer Linie generieren und mit einer weiteren anzeigen. Als Argumente gebt Ihr erst alle x- und dann alle y-Koordinaten:

```python
plt.plot(x, y)
plt.show()
```

Um mehrere Linien anzuzeigen, könnt ihr mehrmals plotten, bevor ihr plt.show() aufruft:

```python
plt.plot(x, g(x))
plt.plot(x, h(x))
plt.show()
```

Oft ist es auch sinnvoll, eure Daten als linlog- oder loglog-Plot darzustellen:

```python
# y-Achse logarithmisch
plt.semilogy(x, y)
plt.show()

# x-Achse logarithmisch
plt.semilogx(x, h(x))
plt.show()

# Beide Achsen logarithmisch
plt.loglog(x, y)
plt.show()
```

Zu guterletzt könnt ihr eurer Darstellung auch Beschriftungen, Farbe und einen Titel hinzufügen:

```python
plt.title("log vs sin")
plt.xlabel("x")
plt.ylabel("Funktionswert")
plt.plot(x, g(x), color="blue")
plt.plot(x, h(x), color="#B72818")
plt.legend(("sin", "log"))

plt.show()
```
