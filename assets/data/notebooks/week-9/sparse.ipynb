```python
import scipy.sparse as sp
import numpy as np
```

```python
n = 4
v = np.arange(n) + 1
```

## Erstellung von diagonalen
Die folgende Zelle enthät code, mit dem ihr diagonalmatrizen mit mehreren subdiagonalen erstellen könnt:
- das erste Argument beschreibt die Vektoren, die Ihr als diagonale einfügen möchtet
- offsets beschreibt, wie weit neben der Hauptdiagonale die $i$-te subdiagonale ist

```python
sd = sp.diags(
    [v, v], 
    offsets=[0, 2], 
    shape=[n,n], 
    format="coo")
print(sd)
```

## Formate
Je nach Bedürfnis könnt ihr matrizen in unterschiedlichen Formaten speichern. Seht das `format`-argument. Für die Diagonalen passt beispielsweise das `"dia"`-format am besten. 

```python
# andere Formate:
# - csr: Zeilenspeicherung
# - csc: Spaltenspeicherung
# - dia: Diagonalenspeicherung
print(sp.diags(
    [v, v], 
    offsets=[0, 2], 
    shape=[n,n], 
    format="dia"))
```

## Dreiecke
Mit `sp.triu` und `sp.tril` könnt ihr die Dreiecke oberhalb oder unterhalb der Diagonale `k` extrahieren: 

```python
# Hauptdiagonale aufwärts
print(sp.triu(sd, k=0))

# Oberhalb der Hauptdiagonalen (Subdiagonale 1 aufwärts):
print(sp.triu(sd, k=1))

# Haupdiagonale abwärts:
print(sp.tril(sd, k=0))

# Unter der Hauptdiagonalen: (leer)
print(sp.tril(sd, k=-1))
```
