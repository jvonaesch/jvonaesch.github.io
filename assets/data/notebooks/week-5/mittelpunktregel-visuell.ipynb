```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = lambda x: np.sin(x) + 1
```

```python
a, b = 0, 10
n, N = 10, 500
x = np.linspace(a, b, n, endpoint=False) + 0.5*(b-a) / n
X = np.linspace(a, b, N)
y = f(x)
Y = f(X)
```

```python
plt.bar(x, y, width=1, color="#ddd", label="Angenähertes Integral")
plt.plot(X, Y, label=r"$f(x)$")
plt.scatter(x, y, label="Messstellen")
plt.legend()
```
