## Rekursive Berechnung
Die folgende Funktion nimmt einen Vektor von Werten $x$, über die die Wurzel berechnet werden soll, und berechnet iterativ die Wurzel aller $x$ mit dem Newton-Verfahren. Die rekursive Formel lautet:

$y_{i+1}=\frac{y_i+\frac{x}{y_i}}{2}$

Dabei gilt:     $\lim_{i\to\infty}(y_i)=\sqrt{x}$

Besonders ist hier, dass wir die Berechnung für jeden Interationsschritt auf einer separaten Zeile der Matrix schreiben und jeweils für alle $x$ gleichzeitig berechnen. Das lesen von Zeilen (oder Spalten) einer Matrix als Vektoren werdet Ihr in der Aufgabenserie 2 gut gebrauchen können. 

```python
def approx_sqrt(x, d):
    y = np.ones((d, x.shape[0]))
    for i in range(1, d):
        y[i] = (y[i-1] + x/y[i-1]) / 2
    return y
```

In der folgenden Matrix seht ihr dieses Verfahren illustriert. Erinnert euch daran, dass ihr Rekursion über die Zeilen einer Matrix durchführen könnt. 

```python
print(approx_sqrt(np.arange(5)+1, 6))
```

Dieses Vorgehen speichert alle Schritte der Rekursion in einer Zeile. So könnt ihr z.B. alle Rekursionsschritte auf einmal plotten: 

```python
n, d = 10000, 9
x = np.arange(1, n)

# berechne alle Schritte
y = approx_sqrt(x, d)

# plot
plt.plot(x, np.sqrt(x), label=r"$\sqrt{x}$")
for i in range(1, d, 2):
    plt.plot(x, y[i], label=r"$y_{%s}(x)$" % (i))
plt.title(r"graduelle Annäherung an $\sqrt{x}$")
plt.legend()
plt.xlabel(r"$x$")
plt.ylabel(r"$y(x)$")
plt.axis((0, n, 0, np.sqrt(n)))
```
